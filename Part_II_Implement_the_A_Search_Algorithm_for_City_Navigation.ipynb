```python
import heapq

# Graph represented as a dictionary
graph = {
    "Seattle": {
        "San Francisco": 678,
        "Chicago": 1737
    },
    "San Francisco": {
        "Seattle": 678,
        "Los Angeles": 348,
        "Riverside": 386
    },
    "Los Angeles": {
        "San Francisco": 348,
        "Riverside": 50,
        "Phoenix": 357
    },
    "Riverside": {
        "Los Angeles": 50,
        "San Francisco": 386,
        "Phoenix": 307,
        "Chicago": 1704
    },
    "Phoenix": {
        "Los Angeles": 357,
        "Riverside": 307,
        "Dallas": 887,
        "Houston": 1015
    },
    "Dallas": {
        "Phoenix": 887,
        "Houston": 225,
        "Atlanta": 721,
        "Chicago": 805
    },
    "Houston": {
        "Dallas": 225,
        "Atlanta": 702,
        "Miami": 968
    },
    "Chicago": {
        "Seattle": 1737,
        "Riverside": 1704,
        "Dallas": 805,
        "Atlanta": 588,
        "Detroit": 238
    },
    "Detroit": {
        "Chicago": 238,
        "Boston": 613,
        "New York": 482,
        "Washington": 396
    },
    "Atlanta": {
        "Dallas": 721,
        "Houston": 702,
        "Chicago": 588,
        "Washington": 543,
        "Miami": 604
    },
    "Washington": {
        "Detroit": 396,
        "Atlanta": 543,
        "Philadelphia": 123,
        "Miami": 923
    },
    "Philadelphia": {
        "Washington": 123,
        "New York": 81
    },
    "New York": {
        "Philadelphia": 81,
        "Boston": 190,
        "Detroit": 482
    },
    "Boston": {
        "New York": 190,
        "Detroit": 613
    },
    "Miami": {
        "Atlanta": 604,
        "Houston": 968,
        "Washington": 923
    }
}

# Heuristic values h(n) (estimate distance from target to Boston(goal))
heuristic = {
    "Seattle": 2496,
    "San Francisco": 2064,
    "Los Angeles": 1875,
    "Riverside": 1700,
    "Phoenix": 1550,
    "Dallas": 1399,
    "Houston": 1605,
    "Chicago": 983,
    "Detroit": 613,
    "Atlanta": 1102,
    "Washington": 394,
    "Philadelphia": 296,
    "New York": 190,
    "Boston": 0,
    "Miami": 1258
}

import heapq

def a_star(graph, start, goal, heuristic):
    open_set = []
    heapq.heappush(open_set, (heuristic[start], start))

    # Actual cost from start
    g_cost = {city: float('inf') for city in graph}
    g_cost[start] = 0

    # Store parent of each city
    parent = {}
    parent[start] = None

    while open_set:
        current = heapq.heappop(open_set)[1]

        if current == goal:
            # Reconstruct path
            path = []
            city = goal

            while city is not None:
                path.append(city)
                city = parent[city]

            path.reverse()
            return path, g_cost[goal]

        for neighbor, distance in graph[current].items():
            tentative_g = g_cost[current] + distance

            if tentative_g < g_cost[neighbor]:
                g_cost[neighbor] = tentative_g
                parent[neighbor] = current

                f_cost = tentative_g + heuristic[neighbor]
                heapq.heappush(open_set, (f_cost, neighbor))

    return None, float('inf')
```

```python
# User Input
start_city = input("Enter start city: ")
goal_city = input("Enter goal city: ")

path, cost = a_star(graph, start_city, goal_city, heuristic)

if path:
    print("\nShortest Path:")
    print(" -> ".join(path))

    print("\nCities Along the Path:")
    for city in path:
        print(city)

    print("\nDistances Between Cities:")
    for i in range(len(path)-1):
        distance = graph[path[i]][path[i+1]]
        print(f"{path[i]} -> {path[i+1]} = {distance}")

    print("\nTotal Cost:", cost)
else:
    print("No path found.")
```
